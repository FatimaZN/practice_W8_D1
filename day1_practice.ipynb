```python
# Toolkit — run this cell as-is
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

iris = sns.load_dataset("iris")
iris.head()
```

### Warm-up: explore the data

Print the shape, and show how many flowers of each species there are.

```python
# 1) Print the shape of the dataset
# YOUR CODE HERE
print("Shape:", iris.shape)
print()
print(iris.info())

# 2) Count how many flowers of each species (hint: value_counts on the 'species' column)
# YOUR CODE HERE

iris["species"].value_counts()
```

### Step 1 — Features (X) and Target (y)

Put the four measurement columns in `X` and the `species` column in `y`.

```python
# X = the four measurement columns
X = iris[["sepal_length", "sepal_width", "petal_length", "petal_width"]]

# y = the species column
y = iris["species"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

### Step 2 — Train / Test split

Hold out **25%** for testing this time. Use `random_state=42`.

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    # YOUR CODE HERE  (pass X, y, test_size=..., random_state=42)
    X,y, test_size=0.25, random_state=42
)

print("Training examples:", X_train.shape[0])
print("Testing examples: ", X_test.shape[0])
```

### Step 3 — Create and train a KNN model

Use **3** neighbors this time (`n_neighbors=3`), then `.fit()` on the training data.

```python
# Create the model
from sklearn.neighbors import KNeighborsClassifier
model = KNeighborsClassifier(n_neighbors=3)

# Train it
# YOUR CODE HERE
model.fit(X_train, y_train)

print("Trained! ✅")
```

### Step 4 — Predict on the test set

```python
predictions = model.predict(X_test)

compare = pd.DataFrame({
    "predicted": predictions[:10], 
    "actual": y_test.values[:10]
})
compare
```

### Step 5 — Score it

Compute and print the accuracy as a percentage.

```python
acc = accuracy_score(y_test, predictions)
print(f"Accuracy: {acc*100:.0f}%")
```

### Bonus — predict a new flower

Make up one flower's four measurements and predict its species.

```python
new_flower = pd.DataFrame(
    [[6.0, 2.7, 5.1, 1.6]],
    columns=["sepal_length", "sepal_width", "petal_length", "petal_width"]
)

new_flower
guess = model.predict(new_flower)
print("The model thinks ths flower is : ", guess[0])
```
